## 5. Errores Comunes a Evitar

### 🚨 Errores Críticos en Tesis de Deep Learning Financiero

#### 5.1 Look-Ahead Bias
- **Error:** Usar información futura para predecir el pasado
- **Solución:** Implementar strict temporal splitting
- **Validación:** Walk-forward analysis, no cross-validation aleatoria

#### 5.2 Overfitting Temporal
- **Error:** Modelos que memorizan patrones específicos de períodos de entrenamiento
- **Solución:** Validación en múltiples regímenes de mercado
- **Técnica:** Purged GroupTimeSeriesSplit

#### 5.3 Survivorship Bias
- **Error:** Solo analizar WTI, ignorar otros contratos/vencimientos
- **Solución:** Incluir análisis de robustez con Brent, contrato continuo

#### 5.4 Sentiment Leakage
- **Error:** Usar sentimientos que incluyen información post-evento
- **Solución:** Timestamp estricto, lead-lag analysis

#### 5.5 Statistical Significance Misconceptions
- **Error:** Multiple testing sin corrección
- **Solución:** Bonferroni, FDR corrections
- **Técnica:** Bootstrap confidence intervals
